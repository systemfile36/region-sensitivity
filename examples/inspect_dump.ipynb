# Inspect an SSAT dump
Run the CLI quickstart first, then point `dump_path` at the result.

```python
from pathlib import Path
from ssat.core.dump import DumpReader

dump_path = Path('/tmp/ssat-quickstart')
reader = DumpReader(dump_path)
manifest = reader.read_manifest()
manifest.adapter_spec
```

```python
clean = reader.read_clean().to_pandas()
perturbed = reader.read_perturbed().to_pandas()
clean['status'].value_counts(), perturbed['status'].value_counts()
```
